Ejercicio 2 - Selección de Personal Estricta. En este ejercicio, la equivalencia biológica exige que el cromosoma demuestre exactamente 
5 rasgos (candidatos). La aptitud se modifica para poder penalizar a los cromosomas que no cumplan con esta suma exacta de bits encendidos o activos


Se tienen 12 candidatos para conformar un equipo de desarrollo.
Cada candidato posee una puntuación de habilidad técnica.

El cromosoma binario representa la selección:
    1 = candidato seleccionado
    0 = candidato no seleccionado

Restricción:
    El equipo debe estar conformado exactamente por 5 personas.

Si el cromosoma no contiene exactamente 5 bits encendidos,
se aplica una penalización a la aptitud.


```python
#Importación de librerias 

import random

random.randint(0,1)

#Habilidades de los candidatos 
habilidades =[85, 72, 90, 65, 88, 76,
    95, 70, 82, 91, 68, 87
    ]
#Número de persona requeridas 
NUM_PERSONAS = 5

#Penalización 
PENALIZACION=100
#Funcón principal 
def decodificar_y_evaluar_e2(cromosoma):
    personas_seleccionada = sum(cromosoma)
    habilidad_total=sum(
        habilidades[i] * gen
        for i, gen in enumerate(cromosoma)
    
    )
    if (personas_seleccionada !=  NUM_PERSONAS and personas_seleccionada <= NUM_PERSONAS) :
        diferencia= abs(personas_seleccionada - NUM_PERSONAS)
        penalizacion = diferencia * PENALIZACION
        return max(1, habilidad_total-penalizacion)
    return habilidad_total
cromosoma_ej =[random.randint(0,1)
    for _ in range(12)
]
aptitud=decodificar_y_evaluar_e2(cromosoma_ej)

print(f"Cromosoma:{cromosoma_ej}")
print (f"Personas seleccionadas: {sum(cromosoma_ej)}")
print(f"Aptitudes: {aptitud}")
```
