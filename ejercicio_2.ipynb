Ejercicio 2 - Selección de Personal Estricta. En este ejercicio, la equivalencia biológica exige que el cromosoma demuestre exactamente 
5 rasgos (candidatos). La aptitud se modifica para poder penalizar a los cromosomas que no cumplan con esta suma exacta de bits encendidos o activos

```python
import random

# Habilidad tecnica de los 12 candidatos disponibles
habilidades = [85, 90, 78, 92, 88, 76, 95, 89, 81, 93, 87, 80]
CANDIDATOS_REQUERIDOS = 5

def decodificar_y_evaluar_e2(cromosoma):
    personal_seleccionado = sum(cromosoma)
    habilidad_total = sum(habilidades[i] * gen for i, gen in enumerate(cromosoma))
    
    # Penalizar si el número de bits en 1 es distinto a 5
    if personal_seleccionado != CANDIDATOS_REQUERIDOS:
        diferencia = abs(personal_seleccionado - CANDIDATOS_REQUERIDOS)
        penalizacion = diferencia * 100
        return max(0, habilidad_total - penalizacion),
        
    return habilidad_total,

# Ejemplo con un cromosoma binario de 12 bits
cromosoma_ejemplo_2 = [random.randint(0, 1) for _ in range(12)]
print(f"Cromosoma: {cromosoma_ejemplo_2} | Aptitud: {decodificar_y_evaluar_e2(cromosoma_ejemplo_2)}")
```
