Ejercicio 1 - Portafolio de Inversiones. En este ejercicio, se plantea la funcion de mapeo en donde este es el que traduce el genotipo binario al espacio de busquedade busqueda de los proyectos, llegando a evuluar el costo y el retorno de la inversion. En donde si el cromosoma excede el presupuesto, la formula matematica aplica la penalizacion establecida

```python
import random

# Datos de los 10 proyectos posibles (Costo, Retorno)
proyectos = [
    (10, 15), (20, 35), (15, 20), (30, 45), (25, 30),
    (40, 50), (10, 25), (35, 60), (20, 40), (50, 70)

]

PREMAXIMO = 100

def decodificar_y_evaluar_e1(cromosoma):
    costo_total = sum(proyectos[i][0] * gen for i, gen in enumerate(cromosoma))
    retorno_total = sum(proyectos[i][1] * gen for i, gen in enumerate(cromosoma))
    
    # Penalización fuerte si se excede el limite del presupuesto máximo
    if costo_total > PREMAXIMO:
        penalizacion = (costo_total - PREMAXIMO) * 10
        return max(0, retorno_total - penalizacion), # Retorna tupla 
        
    return retorno_total,

# Ejemplo de cromosoma binario de 10 bits
cromosoma_ej = [random.randint(0, 1) for _ in range(10)]
print(f"Cromosoma: {cromosoma_ej} | Aptitud: {decodificar_y_evaluar_e1(cromosoma_ej)}")
```
