Ejercicio 3 - Operador de Cruzamiento de Dos Puntos: Modifiquen la función de cruzamiento estándar (que usa un solo punto de corte) para implementar un cruzamiento de dos puntos. Se deben seleccionar dos índices aleatorios y los hijos intercambian el segmento central. 

Definicion de variables y importación de líbrerias 


```python
import random

def cruzamiento_dos_puntos(padre1, padre2):

    # Seleccionar dos puntos de corte aleatorios
    punto1, punto2 = sorted(
        random.sample(range(1, len(padre1)), 2)
    )

    # Crear hijos intercambiando el segmento central
    hijo1 = (
        padre1[:punto1]
        + padre2[punto1:punto2]
        + padre1[punto2:]
    )

    hijo2 = (
        padre2[:punto1]
        + padre1[punto1:punto2]
        + padre2[punto2:]
    )
    return hijo1, hijo2
padre1 = [1, 0, 1, 1, 1, 0, 0, 1, 0, 1]
padre2 = [0, 1, 0, 0, 0, 1, 1, 0, 1, 0]

hijo1, hijo2 = cruzamiento_dos_puntos(padre1, padre2)

print("Padre 1:", padre1)
print("Padre 2:", padre2)
print("Hijo 1: ", hijo1)
print("Hijo 2: ", hijo2)
```
